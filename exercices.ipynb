# **1.4 continue, break et pass**

```python
mesures = [12, None, 18, -999, 25]
retenues = []

for mesure in mesures:
  if mesure is None:
   continue
  if mesure == -999:
    break
  retenues.append(mesure)

print(retenues)
```

# **4.3 Une fonction de moyenne avec un cas limite**

```python
def moyenne(valeurs):
  ###Renvoie la moyenne d'une liste non vide de nombre.###
  if not valeurs:
    raise ValueError("La liste ne doit pas etre vide")
  return sum(valeurs) / len(valeurs)

print(moyenne([10, 20, 30]))
```

# **4.4 Renvoyer plusieurs informations**

```python
def resumer(valeurs):
  if not valeurs:
    raise ValueError("La liste ne doit pas etre vide")
  return len(valeurs), min(valeurs), max(valeurs)

nombre, minimum, maximum = resumer([8, 3, 12])
print(nombre, minimum, maximum)

print(resumer([10, 20, 30]))
#
```

```python
def moyenne_valide(valeurs):
  valides = [x for x in valeurs if x is not None and x >= 0]
  if not valides:
    raise ValueError("La liste ne doit pas etre vide")
  return sum(valides) / len(valides)

print(moyenne_valide([10, None, -5, 20, 30]))

try:
  print(moyenne_valide([None, -2]))
except ValueError as e:
  print("Erreur :", e)
```
