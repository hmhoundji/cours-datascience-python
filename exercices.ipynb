# **if elif** **else**

```python
temperature = 28
if temperature < 0:
    categorie = "gel"
elif temperature < 25:
    categorie ="moderée"
else:
    categorie ="élevée"

print(categorie)
```

```python
mesure = 150

if mesure is None :
  print("mesure manquante")
elif 0 <= mesure <= 100:
  print("mesure valide")
else:
  print("mesure hors limite")
```
